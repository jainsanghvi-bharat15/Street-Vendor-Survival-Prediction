## Missing Value Analysis

This section identifies missing values in each feature and calculates their percentage in the dataset.

```python
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Column": missing.index,
    "Missing Values": missing.values,
    "Percentage (%)": ((missing.values / len(df)) * 100).round(2)
})

missing_df.sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=missing_df,
    x="Column",
    y="Missing Values")

plt.xticks(rotation=90)
plt.title("Missing Values in Dataset")
plt.tight_layout()
plt.savefig(os.path.join(figure_path, "missing_values.png"), dpi=300)
plt.show()
```

## Duplicate Record Analysis

```python
print("Duplicate Records :", df.duplicated().sum())
```

```python
target_counts = df["vendor_survived"].value_counts()
target_counts
```

```python
plt.figure(figsize=(6,5))
sns.countplot(
    data=df,
    x="vendor_survived")
plt.title("Vendor Survival Distribution")
plt.savefig(
    os.path.join(figure_path, "vendor_survival_distribution.png"), dpi=300)
plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.pie(
    target_counts,
    labels=["Active","Closed"],
    autopct="%1.1f%%",
    startangle=90)

plt.title("Vendor Survival Percentage")
plt.savefig(
    os.path.join(figure_path, "vendor_survival_pie.png"), dpi=300)
plt.show()
```

## Numerical Feature Analysis

```python
numerical_columns = [
    "vendor_age_years",
    "years_in_business",
    "avg_daily_revenue_inr",
    "avg_daily_customers",
    "monthly_stall_rent_inr",
    "num_helpers",
    "hours_open_per_day",
    "competition_within_100m",
    "monthly_health_inspection_score",
    "avg_monthly_rainfall_mm",
    "customer_complaint_rate"
]
```

```python
for column in numerical_columns:
    plt.figure(figsize=(8,5))
    sns.histplot(
        df[column],
        kde=True,
        bins=30)
    plt.title(f"Distribution of {column}")
    plt.tight_layout()
    plt.savefig(
        os.path.join(figure_path, f"{column}_distribution.png"), dpi=300)
    plt.show()
```

```python
for column in numerical_columns:
    plt.figure(figsize=(8,3))
    sns.boxplot(x=df[column])
    plt.title(f"Boxplot of {column}")
    plt.tight_layout()
    plt.savefig(os.path.join(figure_path, f"{column}_boxplot.png"), dpi=300)
    plt.show()
```

## Skewness Analysis

```python
skew_df = pd.DataFrame({
    "Feature": numerical_columns,
    "Skewness": [
        round(df[col].skew(),2)
        for col in numerical_columns]
})
skew_df
```

## Correlation Analysis

```python
corr = df[numerical_columns + ["vendor_survived"]].corr()
plt.figure(figsize=(12,8))
sns.heatmap(
    corr,
    annot=True,
    cmap="RdYlGn",
    fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(figure_path, "correlation_heatmap.png"), dpi=300)
plt.show()
```

## Relationship Between Numerical Features and Vendor Survival

```python
for column in numerical_columns:
    plt.figure(figsize=(8,5))
    sns.boxplot(
        data=df,
        x="vendor_survived",
        y=column)
    plt.title(f"{column} vs Vendor Survival")
    plt.tight_layout()
    plt.savefig(os.path.join(figure_path, f"{column}_vs_survival.png"), dpi=300)
    plt.show()
```
